HYPOTHESIS 1: Technology products have the highest profit margin

```python
df.columns
```

```python
cat_profit=df.groupby("Category")["Profit"].sum()
cat_profit.plot(kind='bar')
plt.title("Total profit by category")
plt.xlabel("Category")
plt.ylabel("Profit")
plt.show()
#conclusion= The hypothesis is right
```

HYPOTHESIS 2: The east region has the highest sales compared to the other regions

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("superstore_dataset2011-2015.csv")  

reg_sales = df.groupby("Region")["Sales"].sum()
reg_sales.plot(kind="bar")
plt.title("Sales by region")
plt.xlabel("Region")
plt.ylabel("Sales")
plt.show()
```

HYPOTHESIS 3: Sales are higher during certain months of the year

```python
df.columns
```

```python
df["Order Month"]=pd.DatetimeIndex(df["Order Date"]).month
month_sales=df.groupby("Order Month")["Sales"].sum()
month_sales.plot(kind="line")
plt.title("Sales by month")
plt.xlabel("Order Month")
plt.ylabel("Sales")
plt.show()
# conclusion: The hypothesis is supported
```

HYPOTHESIS 4:Orders with same day shipping have the lowest rate of returned products

```python
df["Ship Mode"].unique()
```
